```python
import numpy as np
```

```python
class MSE:
    def forward( self , y_true , y_pred ):
        self.y_pred = y_pred
        self.y_true = y_true
        loss = np.mean(( y_pred - y_true )**2)
        return loss

    def backward(self):
        n = self.y_true.shape[0]
        return (2/n) * (self.y_pred - self.y_true)
```

```python
loss = MSE()

y_true1 = np.array([[23]])
y_pred1 = np.array([[27]])

y_true2 = np.array([[1],[0],[1]])
y_pred2 = np.array([[0.8],[0.3],[0.9]])

print(loss.forward(y_true1, y_pred1))
print(loss.backward())

print(loss.forward(y_true2, y_pred2))
print(loss.backward())
```
