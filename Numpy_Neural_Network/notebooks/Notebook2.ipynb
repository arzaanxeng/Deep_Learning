```python
import numpy as np
```

```python
# Here i will be creating the hidden layers which corresponds to Dense() method
```

```python
class Dense:
    def __init__( self , input_size , output_size ):
        self.W = np.random.randn(input_size , output_size)
        self.B = np.random.randn( 1 , output_size)

    def forward( self , X):
        self.X = X

        self.Z = X@self.W + self.B
        return self.Z
```

```python
def sigmoid(Z):
    return 1 / (1 + np.exp(-Z))
```

```python
# Lets take some input and test it
X = np.array([[10,14,18,16]])
layer1 = Dense(4,5)
Z1 = layer1.forward(X)
A1 = sigmoid(Z1)
# Passing the output of layer1 to layer2
layer2 = Dense(5,2)
Z2 = layer2.forward(A1)
A2 = sigmoid(Z2)

# Passing the output of layer2 to output layer
output_layer = Dense(2,1)
Z3 = output_layer.forward(A2)
A3 = sigmoid(Z3)
```
